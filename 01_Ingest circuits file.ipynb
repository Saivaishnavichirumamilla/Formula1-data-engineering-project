```python
df_circuits= spark.read \
    .options(header='true') \
    .csv("abfss://formula1@formula1adlssai.dfs.core.windows.net/raw/circuits.csv")
    
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType
```

```python
circuits_schema= StructType(fields=[StructField("circuitId",IntegerType(),False),
                                   StructField("circuitRef",StringType(),True),
                                   StructField("name",StringType(),True),
                                   StructField("location",StringType(),True),
                                   StructField("country",StringType(),True),
                                   StructField("lat",DoubleType(),True),
                                   StructField("lng",DoubleType(),True),
                                   StructField("alt",IntegerType(),True),
                                   StructField("url",StringType(),True)])
```

```python
df_circuits_evaluted_schema= spark.read\
    .schema(circuits_schema)\
    .options(header='true') \
    .csv("abfss://formula1@formula1adlssai.dfs.core.windows.net/raw/circuits.csv")
```

```python
display(df_circuits_evaluted_schema)
```

```python
from pyspark.sql.functions import col
```

```python
df_circuits_processed = df_circuits_evaluted_schema\
                         .select(col("circuitId").alias("circuit_id"),
                                 col("circuitRef").alias("circuit_ref"),
                                 col("name"),col("location"),col("country"),
                                 col("lat").alias("latitude"),
                                 col("lng").alias("longitude"),
                                 col("alt").alias("altitude"))
                    
display(df_circuits_processed)
```

```python
from pyspark.sql.functions import current_timestamp
df_circuits_processed = df_circuits_processed.withColumn("ingestion_date",current_timestamp())
```

```python
df_circuits_processed.write.mode("overwrite").format("parquet").save("abfss://formula1@formula1adlssai.dfs.core.windows.net/processed/circuits/")
```
